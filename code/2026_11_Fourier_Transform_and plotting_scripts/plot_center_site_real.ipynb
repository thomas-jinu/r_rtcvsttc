# Plot center-site real data

Load `collected_operator_data.csv` produced by `process_twotime.py`, select center-site to center-site records, and plot the real component against frequency. Update the path and `CENTER_SITE` below if needed.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

RUN_FOLDER = Path('/Users/qqt/Library/CloudStorage/OneDrive-OakRidgeNationalLaboratory/Research_Projects/r_stcvsttc/code/template_folder_hubbard_chain_cv/N=8_Nup=4_Ndown=4_20261023_140645/G<_singletime_cv_N=8_Nup=4_Ndown=4_20261023_140645_steps=0')
CSV_FILE = RUN_FOLDER / 'collected_operator_data.csv'
CENTER_SITE = 3

if not CSV_FILE.is_file():
    raise FileNotFoundError(f'Data CSV not found: {CSV_FILE}. Run process_twotime.py first.')
data = pd.read_csv(CSV_FILE)
required = {'omega', 'center_site', 'site_index', 'real_part'}
missing = required.difference(data.columns)
if missing:
    raise ValueError(f'CSV is missing required columns: {sorted(missing)}')

center_data = data.loc[
    (data['center_site'] == CENTER_SITE) & (data['site_index'] == CENTER_SITE)
].sort_values('omega')
if center_data.empty:
    raise ValueError(f'No center-site to center-site records for site {CENTER_SITE}.')
center_data.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(center_data['omega'], center_data['real_part'], marker='o', markersize=3, linewidth=1)
ax.set_xlabel('Frequency (ω)')
ax.set_ylabel('Real part')
ax.set_title(f'Center-site to center-site real value (site {CENTER_SITE})')
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```
